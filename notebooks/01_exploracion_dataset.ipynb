```python
import pandas as pd
from pathlib import Path

CSV_DIR = Path("../csv")

mass_train = pd.read_csv(CSV_DIR / "mass_case_description_train_set.csv")
mass_test = pd.read_csv(CSV_DIR / "mass_case_description_test_set.csv")
calc_train = pd.read_csv(CSV_DIR / "calc_case_description_train_set.csv")
calc_test = pd.read_csv(CSV_DIR / "calc_case_description_test_set.csv")

print("MASS TRAIN")
display(mass_train[["patient_id", "pathology", "image file path", "cropped image file path", "ROI mask file path"]].head())

print("MASS TEST")
display(mass_test[["patient_id", "pathology", "image file path", "cropped image file path", "ROI mask file path"]].head())

print("CALC TRAIN")
display(calc_train[["patient_id", "pathology", "image file path", "cropped image file path", "ROI mask file path"]].head())

print("CALC TEST")
display(calc_test[["patient_id", "pathology", "image file path", "cropped image file path", "ROI mask file path"]].head())
```

```python
print("Distribución MASS TRAIN")
print(mass_train["pathology"].value_counts())

print("\nDistribución MASS TEST")
print(mass_test["pathology"].value_counts())

print("\nDistribución CALC TRAIN")
print(calc_train["pathology"].value_counts())

print("\nDistribución CALC TEST")
print(calc_test["pathology"].value_counts())
```
